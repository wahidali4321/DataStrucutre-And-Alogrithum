### Move All Zeros to End

```python
arr = [0, 5, 0, 3, 8, 0, 2, 0, 7]

index = 0
for x in arr:
    if x != 0:
        arr[index] = x
        index += 1

while index < len(arr):
    arr[index] = 0
    index += 1

print(arr)
```

## Find Duplicate

```python
arr = [1, 3, 4, 2, 5, 3, 6]

seen = []

for x in arr:

    if x in seen:
        print("Duplicate:", x)
        break

    seen.append(x)
```
